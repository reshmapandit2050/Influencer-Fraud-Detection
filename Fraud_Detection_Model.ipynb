# **Create Fake vs Real Influencer Label**

```python
# Create authenticity label
df['authenticity'] = df.apply(
    lambda row: 'Fake' 
    if row['engagement_rate'] < 0.03 and row['followers_count'] > 100000 
    else 'Real',
    axis=1
)
```

```python
# Create Influencer Size Category

def influencer_size(followers):
    if followers < 10000:
        return 'Nano'
    elif followers < 100000:
        return 'Micro'
    elif followers < 500000:
        return 'Macro'
    else:
        return 'Mega'

df['influencer_size'] = df['followers_count'].apply(influencer_size)

df[['followers_count', 'influencer_size']].head()
```

```python
# Engagement Level Category

def engagement_level(rate):
    if rate < 0.02:
        return 'Low'
    elif rate < 0.05:
        return 'Medium'
    else:
        return 'High'

df['engagement_level'] = df['engagement_rate'].apply(engagement_level)
```

# **Select Numeric Features**

```python
# Select numeric features
numeric_features = ['followers_count', 'following_count', 'total_posts', 'avg_likes', 'avg_comments', 'engagement_rate',
                   'growth_rate', 'comment_like_ratio', 'follow_ratio']
```

# **Apply Scaling**

```python
# Apply Scaling
scaler = StandardScaler()

df_scaled = scaler.fit_transform(df[numeric_features])
```

```python
# convert in DataFrame
df_scaled = pd.DataFrame(df_scaled, columns=numeric_features)
```

```python
# 
iso_model.fit(df_scaled)
```

# **Model Building(Anomaly Detection)**

```python
# Prepare model data
x = df_scaled
```

```python
x = df[numeric_features]
```

# **Train Isolation Forest Model**

```python
iso_model = IsolationForest(
    n_estimators=100,
    contamination=0.05,
    random_state=42
)
df['iso_anomaly'] = iso_model.fit_predict(x)
```

# **Train Local Outlier Factor (LOF)**

```python
lof_model = LocalOutlierFactor(
    n_neighbors= 20,
    contamination=0.05
)
df['lof_anomaly'] = lof_model.fit_predict(x)
```

```python
# Flag influencer as fraud (1) if either Isolation Forest or LOF detects an anomaly
df['fraud_flag'] = ((df['iso_anomaly'] == -1) | (df['lof_anomaly'] == -1)).astype(int)
```

# **Fraud Risk Scoring**

# **Convert Anomaly Results Into 0–100 Fraud Score**

```python
df['fraud_risk_score'] = df['fraud_flag'] * 100
```

```python
# Create fraud risk score
df['fraud_risk_score'] = (
    (df['iso_anomaly'] == -1).astype(int)*50 +
    (df['lof_anomaly'] == -1).astype(int)*50 
)
```

# **Classify Influencers: Genuine, Suspicious, High Risk**

```python
# Classify Influencers
def classify_influencer(score):
    if score >= 80:
        return "High Risk"
    elif score >= 40:
        return "Suspicious"
    else:
        return "Genuine"
        
df['influencer_type'] = df['fraud_risk_score'].apply(classify_influencer)        
```
